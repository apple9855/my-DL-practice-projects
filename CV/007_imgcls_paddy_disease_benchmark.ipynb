## Define training functions for automation

quick train with fine-tune:

```python
def train_simple(arch, item, batch, epochs=3):
    dls = ImageDataLoaders.from_folder(train_path, seed=44, valid_pct=0.2,item_tfms=item, batch_tfms=batch)
    learn = vision_learner(dls, arch, metrics=error_rate)
    learn.fine_tune(epochs, 0.01)
    return learn
```

Add 'gradient accumulation' `accum=1` and 'fine-tune' option:

```python
def train_accum(arch, item, batch, epochs=3, accum=1, finetune=True):
    dls = ImageDataLoaders.from_folder(train_path, seed=44, valid_pct=0.2, item_tfms=item, batch_tfms=batch, bs = 64//accum)
    cbs = GradientAccumulation(64) if accum else []
    learn = vision_learner(dls, arch, cbs=cbs, metrics=error_rate)
    if finetune:
        learn.fine_tune(epochs, 0.01)
        #learn.tta(dl = dls.test_dl(test_file))
    else:
        learn.unfreeze()
        learn.fit_one_cycle(epochs, 0.01)
    return learn
```

```python
def train_pytorch(arch, item, batch, epochs=3):
    dls = ImageDataLoaders.from_folder(train_path, seed=44, valid_pct=0.2,
                                       item_tfms=item, batch_tfms=custom_transforms)
    learn = vision_learner(dls, 
                           arch=models.resnet50, 
                           pretrained=True, 
                           weights=ResNet50_Weights.IMAGENET1K_V2, 
                           metrics=error_rate)
    learn.fine_tune(epochs, 0.01)
    return learn
```

```python
def get_submission(model, dls, file_name, tta=False):
    test_dls = dls.test_dl(test_file)
    if tta:
        p,_ = model.tta(dl=test_dls, use_max=True)
        idxs = p.argmax(dim=1)
    else:
        p,_,idxs = model.get_preds(dl=test_dls, with_decoded=True)
    results = pd.Series(to_np(idxs)).map(classes)
    submit = pd.read_csv(path/'sample_submission.csv')
    submit['label'] = results
    return submit.to_csv(path/file_name, index=False)
```

## Ensemble Model (Swin+ViT)

```python
res_ens = []
#swin results
res_swin= learn5.tta(dl=learn5.dls.test_dl(test_file), use_max=True) 
#res_swin = (pred1, targ1),(pred2, targ2),...
res_ens.append(res_swin)

#vit results
res_vit = learn6.tta(dl=learn6.dls.test_dl(test_file),use_max=True)
res_ens.append(res_vit)

#select the preds only
tta_preds=first(zip(*res_ens)) # tta_pred= (swin-pred1, swin-pred2...),(vit-pred1,vit-pred2...)

#average swin-preds & vit_preds
avg_pr = torch.stack(tta_preds).mean(0)
avg_pr.shape
```

```python
idxs = avg_pr.argmax(dim=1)
results = pd.Series(to_np(idxs)).map(classes)
submit=pd.read_csv(path/'sample_submission.csv')
submit['label'] = results
submit.to_csv(path/'sub_env.csv', index=False)
```

- <u>SWAG pretrained</u> means models are pre-trained with SWAG (Stochastic Weight Averaging-Gaussian) on large datasets (e.g., ImageNet22K), then fine-tuned on ImageNet1K.
- <u>Valid TTA Error</u> refers to error rate on validation set using Test-Time Augmentation (TTA).
- ✅ The *Ensemble model* averages predictions from the two best-performing single models: ⭐️SwinV2-T and ⭐️ViT-B/16. (**Best score 0.97923**)
  
- 🤔 <u>Why RegNet's highest accuracy on ImageNet doesn't guarantee top performance here?</u>

    Although **RegNetY-16GF** achieves the **highest Top-5 accuracy on ImageNet**, it does **not outperform ViT or Swin** on the Paddy Disease dataset. Possible reasons include:
    
    - **Domain Gap**  
      The ImageNet1K pretraining focuses on general categories (e.g., animals, objects), while the Paddy dataset involves fine-grained visual cues like **color variations, textures, and leaf spots**.  
      These visual patterns are quite different from the semantic features RegNet was optimized for.
    
    - **Architectural Sensitivity to Visual Detail**  
      - **Vision Transformers (ViT) and Swin Transformers** benefit from **attention mechanisms** that dynamically capture spatial dependencies and fine-grained patterns.  
        This makes them particularly suited for **localized and subtle variations**, such as plant disease symptoms.
      - **RegNet**, though powerful, relies mainly on fixed-size **convolutional receptive fields**. Even with SE modules, it may not capture **micro-patterns or irregular textures** as effectively as transformer-based architectures.

## Conclusion

This notebook benchmarks a wide range of computer vision models on a real-world leaf disease classification task. It demonstrates:

- ResNet and EfficientNet provide solid baselines with relatively low model size and decent accuracy.
- Swin Transformer and ViT, although larger in size, consistently outperform CNNs on fine-grained leaf disease classification.
- RegNetY (SWAG) has the highest ImageNet Top-5 score, but domain-specific architectures like ViT/Swin performed better on Paddy dataset.
- Using an ensemble of Swin + ViT further boosted performance to **0.97923**, demonstrating the power of combining spatial attention with transformer-level features.
